```python
senate_path = "../data/raw/mit_election_lab/us_senate/8.0/1976-2024-senate-state.csv"
```

```python
duckdb.sql(f"""
SELECT *
FROM read_csv_auto('{senate_path}')
LIMIT 10
""").show()
```

```python
duckdb.sql(f"""
DESCRIBE SELECT *
FROM read_csv_auto('{senate_path}')
""").show()
```

```python
duckdb.sql(f"""
CREATE OR REPLACE VIEW senate AS
SELECT *
FROM read_csv_auto('{senate_path}')
""")
```

```python
duckdb.sql("""
DESCRIBE senate
""").show()
```
