```python
import duckdb
```

```python
house_path = "../data/raw/mit_election_lab/us_house/15.0/1976-2024-house.tab"
```

```python
duckdb.sql(f"""
    SELECT
        candidate,
        party,
        writein,
        fusion_ticket,
        candidatevotes,
        totalvotes
    FROM read_csv_auto(
        '{house_path}',
        types={{'runoff': 'VARCHAR'}}
    )
    WHERE year = 2024
      AND state = 'MAINE'
      AND district = 1
      AND stage = 'GEN'
    ORDER BY candidatevotes DESC
""").show()
```
